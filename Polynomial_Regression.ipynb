# Polynomial Regression

Simply put, polynomial regression is used when linear regression isn't the right model for your data, and it seems as though a curve may be a better fit. In some cases, this may be obvious once you have plotted the data, as in the example used below or it may have seemed initially like linear regression was sufficient.

There are many examples of polynomials in maths.
For example, y = x^3 + 3x^2 + 7x + 5 is a polynomial.


The equation used in polynomial regression is:


y = B_0 + B_1 * X_1 + B_2 * X_2^2 + ... B_n * X_n^n

When you compare this with the equation used in simple linear regression, you can see how simple linear regression is a type of first order polynomial.

Polynomial regression is harder to carry out than linear regression for several reasons. One of the first questions you have to ask yourself when looking at the data you're working with, is what degree of polynomial you're going to need.
The simplest way is simply plotting your data and guessing the degree of polynomial. Then based on the results and or other methods (we'll see that a bit later), you can decide whether to change the degree.

The data used in this example has been made up. x2 + x + 10

```python
# This code shows data with a linear regression model.
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import numpy as np

X = np.array([7.45, 2.31, 8.67, 5.24, 8.23, 4.93, 10.22])
y = np.array([72,18,94,42,86,39,124])
X = X.reshape(-1, 1)

m = LinearRegression()
m.fit(X, y)

print("R²:", m.score(X, y))
plt.scatter(X, y, color = "#a7beae")
y_pred = m.predict(X)
plt.plot(X, y_pred, color = "#b85042")

plt.xlabel("X")
plt.ylabel("y")
plt.grid()
plt.show()
```

Although in this case, it may be hard to tell from the graph and from the R^2 value, there may be a better model than linear regression.

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import numpy as np

X = np.array([7.45, 2.31, 8.67, 5.24, 8.23, 4.93, 10.22])
y = np.array([72,18,94,42,86,39,124])

X = X.reshape(-1, 1)

# This is where the polynomial regression first comes in. Here you specificy the degree of polynomial.
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

# This is similar to the linear regression above, except that X_poly and y is used instead of X and y
m = LinearRegression()
m.fit(X_poly, y)
# Calculating the R^2 value
print("R²:", m.score(X_poly, y))

# This creates a smooth curve
X_plot = np.linspace(X.min(), X.max(), 200).reshape(-1, 1)
X_plot_poly = poly.transform(X_plot)
y_plot = m.predict(X_plot_poly)

# Plot
plt.scatter(X, y, color="#a7beae")
plt.plot(X_plot, y_plot, color="#1e2761", label="Polynomial")




# This is just the linear model to show you the difference
l = LinearRegression()
l.fit(X,y)
y_pred = l.predict(X)
plt.plot(X, y_pred, color = "#b85042", label="Linear")

plt.xlabel("X")
plt.ylabel("y")
plt.grid()
plt.legend()
plt.show()

```

As you can see, the polynomial model was much better, even though it wasn't particulary obvious at the start.

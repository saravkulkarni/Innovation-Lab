# Matplotlib

Matplotlib is a python library used for graph plotting. Reading the comments in the example below will help you understand some of the most important features of this library. The data from this example has come from a database containg information on Cepheid Variables. Clicking on this link will take you to their website where you can find background information and more data: https://www.astro.utoronto.ca/DDO/research/cepheids/cepheids.html#files.

A full reference is given at the bottom of the page.

```python
# The following two lines import both matplotlib and numpy if they aren't already there.
import matplotlib.pyplot as plt
import numpy as np

# The following two lines are the values plotted on the x and y axes.
xpoints = np.array([7.023958, 17.140849, 13.754707, 6.807055, 3.728190, 10.144698])
ypoints = np.array([54.7, 113.8 , 95.8 , 45.5, 35.1, 61.6])

# The next line plots the points as a scatter graph and provides the colour for the points.
plt.scatter(xpoints, ypoints, color = "#b85042")
# Title of the graph
plt.title("Radius vs Period of Cepheid Variables")
# x-axis title
plt.xlabel("Period")
# y-axis title
plt.ylabel("Radius")
# The next lines adds grid lines.
plt.grid()
# This draws the graph.
plt.show()
```

# Data Cleaning

Data cleaning is an important part of the process, as it impacts the quality of data. It is often said that the higher quality your data, the higher quality your analysis will be. Failing to prepare your data may lead to errors and inconsistencies, which may result in flawed outputs and conclusions.

In the following examples, data has been taken from the NASA Exoplanet Archive. Clicking on the following link will take you the webpage where you can see all the data in the set, as only a subsection has been used here: https://exoplanetarchive.ipac.caltech.edu/cgi-bin/TblView/nph-tblView?app=ExoTbls&config=STELLARHOSTS.

More information about what the data in each column means is given here.

sy_name = System Name

hostname = Host Name

sy_snum = Number of Stars

sy_pnum = Number of Planets

st_refname = Stellar Parameter Reference

st_spectype = Spectral Type

st_teff = Stellar Effective Temperature (K)

st_rad = Stellar Radius (Solar Radius)

st_mass = Stellar Mass (Solar Mass)

st_logg = Stellar Surface Gravity (log10(cm/s**2))

sy_dist = Distance (pc)

```python
import pandas as pd
url = "https://raw.githubusercontent.com/saravkulkarni/Innovation-Lab/refs/heads/main/Data/STELLARHOSTS_2026.09.21_07.45.35.csv"
df = pd.read_csv(url)
# The next line removes all the rows that have null values. As you can see, only two rows are completely full.
# The inplace = True part means that instead of creating a new data frame with the result, the original data frame is changed instead.
df.dropna(inplace = True)
print(df.to_string())
```

```python
import pandas as pd
url = "https://raw.githubusercontent.com/saravkulkarni/Innovation-Lab/refs/heads/main/Data/STELLARHOSTS_2026.09.21_07.45.35.csv"
df = pd.read_csv(url)
# Another way of dealing with missing values it to replace them with something else.
# You could decide to replace the missing values in a given column with the mean.
# The next two code lines do exactly this.
# The mean of the stellar mass is calculated and then that value is used to replace the missing values in that column.
x = df["st_mass"].mean()
df.fillna({"st_mass": x}, inplace=True)
# Just for simplicity's sake, only two columns from the data frame are printed here.
# Compare this result with the original which is in the next code block.
print(df[["sy_name", "st_mass"]].to_string())
```

```python
import pandas as pd
url = "https://raw.githubusercontent.com/saravkulkarni/Innovation-Lab/refs/heads/main/Data/STELLARHOSTS_2026.09.21_07.45.35.csv"
df = pd.read_csv(url)
# Imagine that you've collected a lot of data but you realise you actually don't need all of that right now.
# You may decide to focus on a subsection.
# One way to do that would be to filter based on your requirements.
# In this example, objects with less that two planets are dropped.
for x in df.index:
  if df.loc[x, "sy_pnum"] < 2 :
    df.drop(x, inplace = True)
print(df.to_string())
```

```python
import pandas as pd
url = "https://raw.githubusercontent.com/saravkulkarni/Innovation-Lab/refs/heads/main/Data/STELLARHOSTS_2026.09.21_07.45.35.csv"
df = pd.read_csv(url)
# You may find that some rows are duplicates. With large datasets this can be hard to find.
# The next line checks to see if each row is a duplicate.
# In this case, none of the rows are duplicates (they all give False).
print(df.duplicated())
# If you did have duplicates, the next line would remove them.
df.drop_duplicates(inplace = True)
```

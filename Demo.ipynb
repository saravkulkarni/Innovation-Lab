This is a scatter plot drawn using python. A line of linear regression has been plotted, and the r value printed. The r value tells us whether using a linear regression model is suitable or not. In this case, it is.

```python
import matplotlib.pyplot as plt
from scipy import stats

x = [4,10,7,11,2,4,8,12,7,8,4,3]
y = [32,41,35,41,31,32,36,43,34,35,31,31]

slope, intercept, r, p, std_err = stats.linregress(x, y)

def myfunc(x):
  return slope * x + intercept

mymodel = list(map(myfunc, x))

plt.scatter(x, y)
plt.plot(x, mymodel)
plt.show()

print(r)
```

We can use this to make predictions. For example, according to our model, what is the y-value, when x is 6?

```python
y = myfunc(6)
print(y)
```

# Linear Regression

Linear regression is used to model the relationship between one or more independent variables and a dependent variable using a straight line.




## Simple Linear Regression

Simple linear regression only involves one independent variable. The equation takes the form:

y = mX + C

This is the equation of a straight line.

The data from this example has come from a database containg information on Cepheid Variables. Clicking on this link will take you to their website where you can find background information and more data: https://www.astro.utoronto.ca/DDO/research/cepheids/cepheids.html#files.

A full reference is given at the bottom of the page.

Reading the comments will help you understand the code.

```python
# This line imports the linear regression model from scikit-learn
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import numpy as np

# The following two lines contain the data we are going to use
X = np.array([7.023958, 17.140849, 13.754707, 6.807055, 3.728190, 10.144698])
y = np.array([54.7, 113.8 , 95.8 , 45.5, 35.1, 61.6])
# This next line is important, as it is expected that X is a matrix and not a vector.
# This line reshapes it.
X = X.reshape(-1, 1)
# The next two lines fit the data to the model
m = LinearRegression()
m.fit(X, y)
# The R² value is an important indicator of how well the model fits the data.
# It takes a value between 0 and 1, with 1 meaning a perfect fit.
print("R²:", m.score(X, y))
# This gives the gradient/slope of the line.
print("Slope:", m.coef_[0])
# This gives the y-intercept of the line.
print("Intercept:", m.intercept_)
# For our purposes, this next line is the most exciting, as you can make predictions based on the data you have.
# In this case, the radius based on the period.
print("Prediction for 5:", m.predict([[5]])[0])
# These two next lines plot the linear regression line
y_pred = m.predict(X)
plt.plot(X, y_pred, color = "#b85042", linewidth = 3)
# This next line plots the points from the original data
plt.scatter(X, y, color = "#a7beae")
plt.title("Radius vs Period of Cepheid Variables")
plt.xlabel("Period")
plt.ylabel("Radius")
plt.grid()
plt.show()

# Just as a little aside, m.predict() gives an array as an output.
# In this line: print("Prediction for 5:", m.predict([[5]])[0]), [0] has been added
# This is to indicate that we want to print the first item in the array
# We could have put predictions = m.predict([[5],[11],[15]])
# And then either print(predictions)
# Or print(predictions[2]) for example
# y_pred = m.predict(X) creates an array for the predicted examples
```

## Multiple Linear Regression

Multiple linear regression involves, as the name suggests, multiple independent variables. The equation takes the form:

y = B_0 + B_1 * X_1 + B_2 * X_2 + ... B_n * X_n

The data for this example has been made up.




```python
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import numpy as np

X = np.array([
    [5, 21],
    [8, 25],
    [3, 15],
    [7, 26],
    [7, 22],
    [8, 23],
    [6, 22],
    [7, 24],
    [6, 21]
])
# Note that as X is in the required format, we don't need to reshape it.
y = np.array([87, 93, 80, 95, 92, 92, 90, 95, 88])

m = LinearRegression()
m.fit(X, y)
# In this example, there will be two coefficients
print("Coefficients:", m.coef_)
print("Intercept:", m.intercept_)
print("R²:", m.score(X, y))
# The order of the numbers in the next line does matter.
# Keep track of what the variables you're using are and what they mean
# The next line is for a prediction
print("Prediction for 3 and 30:", m.predict([[3,30]])[0])
# As before, the next line creates an array of predicted y-values
y_pred = m.predict(X)
# This plots the acutal values of y against the predicted values of y
# The closer this is to a diagonal line, the more perfect the model
plt.scatter(y, y_pred, color = "#b85042")
plt.title("Actual vs Predicted")
plt.xlabel("Actual")
plt.ylabel("Predicted")
plt.grid()
plt.show()
```
